## 1. Multiply each ComponentX_fraction with its respective properties

````python
for comp_num in range(1, 6):  # For components 1 through 5
    for prop_num in range(1, 11):  # For properties 1 through 10
        # Original column name
        orig_col = f'Component{comp_num}_Property{prop_num}'
        # New column name
        new_col = f'Fraction{comp_num}_Property{prop_num}'
        # Multiply fraction with property and create new column
        train_df[new_col] = train_df[f'Component{comp_num}_fraction'] * train_df[orig_col]
````

````python
def create_features(df):    
    # Volume fractions
    fractions = ['Component1_fraction', 'Component2_fraction', 
                 'Component3_fraction', 'Component4_fraction', 
                 'Component5_fraction']
    # Fraction statistics
    df['max_fraction'] = df[fractions].max(axis=1)
    df['min_fraction'] = df[fractions].min(axis=1)
    df['fraction_range'] = df['max_fraction'] - df['min_fraction']
    
    # Property aggregations
    for prop_idx in range(1, 11):
        prop_cols = [f'Fraction{i}_Property{prop_idx}' for i in range(1,6)]
        df[f'mean_prop_{prop_idx}'] = df[prop_cols].mean(axis=1)
        df[f'std_prop_{prop_idx}'] = df[prop_cols].std(axis=1)
    
    return df
````
